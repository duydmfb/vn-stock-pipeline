### Hành vi khi lỗi

| Trường hợp        | Kết quả                                                            |
|-------------------|--------------------------------------------------------------------|
| Mã sai định dạng (XXXXX) | Ném `ValueError: Invalid symbol...` ngay khi tạo `Quote()`  |
| Mã đúng định dạng, không tồn tại (ZZZ) | Ném `RetryError: RetryError...`               |
| Khoảng toàn ngày nghỉ Tết | Không rỗng, trả về `ngày giao dịch gần đó` (sàn vẫn nghỉ)  |
| Khoảng ngày tương lai | Trả về `10 phiên gần nhất`, kết thúc ở `phiên mới nhất có thật`|

-> API không tự báo lỗi khi khoảng ngày không có phiên giao dịch. Nó âm thầm thay bằng các phiên gần nhất (trong hoặc trước khoảng yêu cầu). Số dòng trả về xấp xỉ số ngày trong khoảng yêu cầu. Pipeline phải tự lọc df theo [start, end] và validator phải chặn ngày tương lai / ngoài khoảng. Chỉ mã sai định dạng mới ném lỗi → cần try/except ValueError khi gọi.

API chậm và có rate limit khi viết lại `d = Quote...`, `d['time']`  pipeline phải (1) lưu dữ liệu thô xuống đĩa để không gọi lại, (2) có retry + chờ tăng dần khi bị chặn. Thoát ra chạy lại từ dầu thì không bị chậm nữa. 

### Suy ra cho pipeline
- Không gọi API dồn dập. Cần nghỉ giữa các lần gọi (rate limiting chủ động).
- Cần retry + chờ tăng dần (exponential backoff), giới hạn số lần thử.
- Cần lưu dữ liệu thô xuống đĩa để không phải gọi lại khi chạy lại transform.

- Nguồn VCI không cắt chặt theo `start`: xin từ 2024-01-01 nhưng trả về từ 2023-11-13 (~1.5 tháng thừa ở đầu). `end` cho đúng ngày nhưng là 0h.
- Kết hợp với các phát hiện trước: KHÔNG được tin biên ngày mà API trả về.


## 4. So sánh nguồn dữ liệu: VCI vs kbs

### Vì sao phải đổi nguồn

Khi viết `main.py` (Ngày 3), xin dữ liệu từ `2024-01-01` với `source='VCI'`
nhưng nhận về bảng bắt đầu từ `2023-11-13` — thừa ~1.5 tháng ở đầu khoảng.
`end` thì được tôn trọng, chỉ `start` bị bỏ qua.

Thử đổi sang `source='kbs'`: xin từ `2026-01-01`, nhận về từ `2026-01-05`
— đúng phiên giao dịch đầu tiên sau ngày yêu cầu (01/01 nghỉ Tết dương,
03–04/01 là cuối tuần).

**→ Chọn `kbs` làm nguồn chính vì cắt đúng biên ngày yêu cầu.**

| Hành vi | VCI | kbs |
|---|---|---|
| Tôn trọng `start` | KHÔNG (trả thừa ~1.5 tháng đầu) | CÓ |
| Tôn trọng `end` | Có | Có |
| Kiểu cột `time` | `datetime64` — date thuần (2023-11-13) | `datetime64` — CÓ giờ (2026-01-05 07:00:00) |

### Vấn đề mới của kbs: cột `time` có phần giờ `07:00:00`

Toàn bộ các dòng đều mang giờ `07:00:00`. Grain của dự án là "1 mã × 1 ngày"
nên phần giờ này là nhiễu — nếu đưa nguyên vào khoá chính sẽ gây rủi ro
trùng lặp / join sai.

→ Bước transform (Ngày 8) BẮT BUỘC chuẩn hoá `time` về date thuần.

### Kiểm tra ngày có bị lệch múi giờ không

Con số `07:00:00` gợi ý có chuyển đổi múi giờ. Cần chắc `2026-01-05 07:00:00`
đúng là phiên ngày 05/01, không phải phiên 04/01 bị đẩy lên.

Cách kiểm chứng: lấy cùng khoảng ngày từ cả hai nguồn, so `close` theo ngày lịch.
